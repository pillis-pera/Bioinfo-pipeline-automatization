**Import relevant libraries**

Note: This Jupyter Notebook was built using Python 3.12.13

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import datetime
import numpy as np
```

**Generate counts matrix from each samples counts vector.**

```python

directory_samples = Path("Counts/")

samples = {}
for file in directory_samples.glob("*.csv"):
    df = pd.read_csv(file, usecols=["Reference","Mapped"])
    sample = file.stem
    samples[sample] = df.set_index("Reference")["Mapped"]
matrix = pd.concat(samples,axis=1)
matrix.index.name = "miRNA"
matrix.to_csv("counts_matrix.csv")
```

**Filter miRNAs that have min counts in x samples**

```python
                    #Filter miRNAs that have:
min_counts = 2      #Atleast this many counts
x_samples = 2       #In this many samples or more

df = pd.read_csv("counts_matrix.csv")
df = df.set_index("miRNA")

x_samples = 1/df.shape[1]
df_filtered = df[(df >= min_counts).mean(axis=1) > x_samples]
df_filtered.to_csv("counts_matrix_filtered.csv")
```
